```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.metrics.pairwise import cosine_similarity
```

```python
import re

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
df["clean_text"] = df["example_user_problem"].apply(clean_text)
df.columns.tolist()
```

```python
X = df["clean_text"]
y_category = df["category"]
y_risk = df["risk_level"]
```

```python
X_train, X_test, y_cat_train, y_cat_test = train_test_split(
    X, y_category, test_size=0.2, random_state=42
)
```

```python
X_train
```

```python
X_test
```

```python
vectorizer = TfidfVectorizer(ngram_range=(1, 2))
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)
```

```python
category_clf = LogisticRegression(max_iter=1000, class_weight="balanced")
category_clf.fit(X_train_vec, y_cat_train)
```

```python
cat_preds = category_clf.predict(X_test_vec)
print(classification_report(y_cat_test, cat_preds, zero_division=0))
```

```python
X_train2, X_test2, y_risk_train, y_risk_test = train_test_split(
    X, y_risk, test_size=0.2, random_state=42
)

X_train2_vec = vectorizer.transform(X_train2)
X_test2_vec = vectorizer.transform(X_test2)

risk_clf = LogisticRegression(max_iter=1000, class_weight="balanced")
risk_clf.fit(X_train2_vec, y_risk_train)

risk_preds = risk_clf.predict(X_test2_vec)
print(classification_report(y_risk_test, risk_preds, zero_division=0))
```

```python
retrieval_vectorizer = TfidfVectorizer(ngram_range=(1, 2))
retrieval_matrix = retrieval_vectorizer.fit_transform(df["clean_text"])
```

```python
def find_similar_case(user_text, top_k=1):
    cleaned = clean_text(user_text)
    query_vec = retrieval_vectorizer.transform([cleaned])
    sims = cosine_similarity(query_vec, retrieval_matrix)[0]
    top_idx = sims.argsort()[::-1][:top_k]
    return df.iloc[top_idx]

# quick test
find_similar_case("my boss hasn't paid me in two months")
```

```python
def predict_triage(user_text, top_k=1):
    cleaned = clean_text(user_text)

    # Category prediction
    cat_vec = vectorizer.transform([cleaned])
    predicted_category = category_clf.predict(cat_vec)[0]

    # Risk prediction
    predicted_risk = risk_clf.predict(cat_vec)[0]

    # Retrieve the closest matching real case
    match = find_similar_case(user_text, top_k=top_k).iloc[0]

    result = {
        "input_text": user_text,
        "predicted_category": predicted_category,
        "predicted_risk_level": predicted_risk,
        "matched_subcategory": match["subcategory"],
        "possible_pathway": match["possible_pathway"],
        "documents_or_evidence": match["documents_or_evidence"],
        "possible_authority": match["possible_authority"],
        "legal_citation": match["legal_citation"],
        "safety_note": match["safety_note"],
    }
    return result

# test it
predict_triage("my boss hasn't paid me in two months")
```

```python
print(predict_triage("someone is stalking me and I'm scared"))
print(predict_triage("my landlord won't return my security deposit"))
print(predict_triage("I want to adopt a child legally"))
```

```python
SAFETY_KEYWORDS = ["hit", "abuse", "threat", "unsafe", "afraid", "violence", "stalk", "assault"]

def add_safety_flag(result):
    text_lower = result["input_text"].lower()
    is_urgent = (
        result["predicted_category"] == "Women's Safety"
        or result["predicted_risk_level"] == "High"
        or any(k in text_lower for k in SAFETY_KEYWORDS)
    )
    result["urgent_flag"] = is_urgent
    return result

predict_triage("my husband hits me and I don't know who to call")
```

```python
import json

with open("corpus.json", "r", encoding="utf-8") as f:
    corpus = json.load(f)

corpus_df = pd.DataFrame(corpus)
corpus_df["clean_text"] = corpus_df["text"].apply(clean_text)
corpus_df.head()
```

```python
statute_vectorizer = TfidfVectorizer(ngram_range=(1, 2))
statute_matrix = statute_vectorizer.fit_transform(corpus_df["clean_text"])

def retrieve_statutes(user_text, category=None, top_k=2):
    cleaned = clean_text(user_text)
    query_vec = statute_vectorizer.transform([cleaned])
    sims = cosine_similarity(query_vec, statute_matrix)[0]

    if category:
        boost = corpus_df["category"].apply(lambda c: 0.15 if c == category else 0)
        sims = sims + boost.values

    top_idx = sims.argsort()[::-1][:top_k]
    return corpus_df.iloc[top_idx][["act", "section", "text"]]

# test it
retrieve_statutes("my husband hits me and threatens me", category="Women's Safety")
```

```python
def predict_triage(user_text, top_k=1):
    cleaned = clean_text(user_text)
    cat_vec = vectorizer.transform([cleaned])
    predicted_category = category_clf.predict(cat_vec)[0]
    predicted_risk = risk_clf.predict(cat_vec)[0]

    match = find_similar_case(user_text, top_k=top_k).iloc[0]
    statutes = retrieve_statutes(user_text, category=predicted_category, top_k=2)

    text_lower = user_text.lower()
    is_urgent = (
        predicted_category == "Women's Safety"
        or predicted_risk == "High"
        or any(k in text_lower for k in SAFETY_KEYWORDS)
    )

    return {
        "predicted_category": predicted_category,
        "predicted_risk_level": predicted_risk,
        "possible_pathway": match["possible_pathway"],
        "documents_or_evidence": match["documents_or_evidence"],
        "possible_authority": match["possible_authority"],
        "legal_citation": match["legal_citation"],
        "retrieved_statutes": statutes.to_dict(orient="records"),
        "safety_note": match["safety_note"],
        "urgent_flag": is_urgent,
    }

predict_triage("my husband hits me and threatens me")
```
